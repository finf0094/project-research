**Objective:** to learn how to create a simple dataset and prepare it for further analysis and machine learning.

## 1. Dataset topic

**Topic: Student performance.** Each record describes one student: study habits, attendance, previous results and life circumstances. The goal is to predict whether the student **passed the final exam**.

## 2–3. The table

The table was created in Excel (`student_performance.xlsx`, sheet *Data*) and contains **55 records**, **8 features** and **1 target variable** (`passed`). `student_id` is only an identifier.

## 4. Column description (data dictionary)

| Column | Data type | Role | Description | Values / range |
|---|---|---|---|---|
| `student_id` | int | Identifier | Unique student number (not a predictive feature) | 1–52 |
| `gender` | string (categorical) | Feature | Student gender | Male / Female |
| `age` | int | Feature | Student age in years | 17–23 |
| `study_hours_per_week` | float | Feature | Average self-study hours per week | 2–25 |
| `attendance_rate` | float | Feature | Share of attended classes, % | 55–100 |
| `previous_grade` | float | Feature | Average grade of the previous semester (0–100) | 35–98 |
| `sleep_hours` | float | Feature | Average hours of sleep per night | 4.5–9.5 |
| `part_time_job` | string (categorical) | Feature | Whether the student has a part-time job | Yes / No |
| `parental_education` | string (categorical) | Feature | Highest education level of parents | School / College / University |
| `passed` | int (binary) | **Target** | Whether the student passed the final exam | 1 = passed, 0 = failed |

The same dictionary is stored on the *Data dictionary* sheet of the Excel file.

## 5. Problematic values

- **7 missing values**: `study_hours_per_week` (2), `attendance_rate` (2), `sleep_hours` (1), `gender` (1), `parental_education` (1);
- **3 duplicate rows** (students 8, 26 and 39 are repeated at the end of the table).

## 10. Conclusion

**What the dataset contains.** 55 records about students with 8 features (gender, age, weekly study hours, attendance rate, previous grade, sleep hours, part-time job, parental education) and a binary target `passed` (1 = passed, 0 = failed). Numeric columns are `age`, `study_hours_per_week`, `attendance_rate`, `previous_grade`, `sleep_hours`; categorical columns are `gender`, `part_time_job`, `parental_education`. The classes are moderately imbalanced (34 passed vs 21 failed).

**Identified problems.**
- 7 missing values in 5 columns (`study_hours_per_week`, `attendance_rate`, `sleep_hours`, `gender`, `parental_education`);
- 3 duplicated rows, which would give some students more weight and could end up in both train and test sets;
- categorical features are stored as text, which most ML algorithms cannot use directly;
- numeric features have very different scales (e.g. sleep 4.5–9.5 vs attendance 55–100);
- `student_id` is an identifier and must not be used as a feature.

**What must be corrected before training.**
1. Remove the duplicate rows (`df.drop_duplicates()`).
2. Fill missing values: median for numeric columns, mode for categorical ones (or drop those rows).
3. Encode categorical variables (binary mapping / one-hot encoding).
4. Drop `student_id`, scale numeric features for distance-based models (KNN, SVM, Logistic Regression).
5. Split the data into training and test sets.
